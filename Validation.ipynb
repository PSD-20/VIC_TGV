```python
import numpy as np


import pickle

with open("data.pkl", "rb") as f:
    data = pickle.load(f)

# Replace these with your own matrices
omega_exact = data["exact"]
omega_num   = data["predicted"]

# Domain
Lx = 2*np.pi
Ly = 2*np.pi

Nx, Ny = omega_exact.shape

dx = Lx/Nx
dy = Ly/Ny
```

```python
def l2_error(exact, numerical):
    return np.sqrt(np.sum((numerical-exact)**2)/np.sum(exact**2))

def relative_error(exact, numerical):
    return np.linalg.norm(numerical-exact)/np.linalg.norm(exact)

def max_vorticity(omega):
    return np.max(np.abs(omega))

def average_vorticity(omega):
    return np.mean(omega)

def circulation(omega, dx, dy):
    return np.sum(omega)*dx*dy

def enstrophy(omega, dx, dy):
    return 0.5*np.sum(omega**2)*dx*dy

def order_of_accuracy(error_coarse, error_fine, h_coarse, h_fine):
    return np.log(error_coarse/error_fine)/np.log(h_coarse/h_fine)
```

```python
L2 = l2_error(omega_exact, omega_num)
Rel = relative_error(omega_exact, omega_num)

max_exact = max_vorticity(omega_exact)
max_num = max_vorticity(omega_num)

avg_exact = average_vorticity(omega_exact)
avg_num = average_vorticity(omega_num)

Gamma_exact = circulation(omega_exact, dx, dy)
Gamma_num = circulation(omega_num, dx, dy)

Ens_exact = enstrophy(omega_exact, dx, dy)
Ens_num = enstrophy(omega_num, dx, dy)

print("="*60)
print("Taylor-Green Validation Report")
print("="*60)
print(f"L2 Error              : {L2:.6e}")
print(f"Relative Error        : {Rel:.6e}")
print()
print(f"Maximum Vorticity")
print(f"  Exact               : {max_exact:.6e}")
print(f"  Numerical           : {max_num:.6e}")
print()
print(f"Average Vorticity")
print(f"  Exact               : {avg_exact:.6e}")
print(f"  Numerical           : {avg_num:.6e}")
print()
print(f"Circulation")
print(f"  Exact               : {Gamma_exact:.6e}")
print(f"  Numerical           : {Gamma_num:.6e}")
print()
print(f"Enstrophy")
print(f"  Exact               : {Ens_exact:.6e}")
print(f"  Numerical           : {Ens_num:.6e}")
print("="*60)
```
